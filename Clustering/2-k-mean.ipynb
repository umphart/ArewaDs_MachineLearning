```python
# %%
# Try Different Clustering Methods - DBSCAN Example
# !pip install scikit-learn matplotlib seaborn

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score

# %%
# Generate synthetic dataset (two interleaving moons)
# This dataset is tricky for K-Means because it's non-spherical
X, y_true = make_moons(n_samples=500, noise=0.07, random_state=42)

plt.scatter(X[:,0], X[:,1], s=30, c=y_true, cmap="viridis")
plt.title("Ground Truth (Two Moons Dataset)")
plt.show()

# %%
# Compare K-Means vs DBSCAN

# --- KMeans ---
kmeans = KMeans(n_clusters=2, random_state=42)
labels_kmeans = kmeans.fit_predict(X)
score_kmeans = silhouette_score(X, labels_kmeans)

# --- DBSCAN ---
dbscan = DBSCAN(eps=0.2, min_samples=5)
labels_dbscan = dbscan.fit_predict(X)
score_dbscan = silhouette_score(X, labels_dbscan)

# %%
# Plot clustering results
fig, ax = plt.subplots(1, 2, figsize=(12,5))

ax[0].scatter(X[:,0], X[:,1], c=labels_kmeans, cmap="viridis", s=30)
ax[0].set_title(f"K-Means Clusters\nSilhouette = {score_kmeans:.2f}")

ax[1].scatter(X[:,0], X[:,1], c=labels_dbscan, cmap="viridis", s=30)
ax[1].set_title(f"DBSCAN Clusters\nSilhouette = {score_dbscan:.2f}")

plt.show()

# %%
# Observations (Markdown cells in Notebook):
# - KMeans fails to capture the "moon shapes", splitting them incorrectly.
# - DBSCAN correctly identifies the two curved clusters.
# - DBSCAN also labels noise points as -1, making it robust to outliers.
# - Lesson: KMeans is good for spherical clusters, but density-based methods like DBSCAN 
#   are better for irregular shapes and uneven densities.
```
