```python
# %%
# Classification Scavenger Hunt - Create a Word Document
# Run this in Jupyter Notebook or VS Code

# Install dependency if not available
# !pip install python-docx

from docx import Document

# %%
# Create a Word document
doc = Document()
doc.add_heading('Classification Algorithms Scavenger Hunt', 0)

doc.add_paragraph(
    "This document explores five classification algorithms from Scikit-learn, "
    "paired with datasets from the curriculum. For each algorithm, we describe "
    "a dataset, a sample question, and why the classification method fits."
)

# %%
# Create a table with header
table = doc.add_table(rows=1, cols=4)
hdr_cells = table.rows[0].cells
hdr_cells[0].text = 'Algorithm'
hdr_cells[1].text = 'Dataset'
hdr_cells[2].text = 'Question'
hdr_cells[3].text = 'How It Works'

# %%
# Fill rows with classification algorithms, datasets, questions, explanations
rows = [
    ("Logistic Regression",
     "Pumpkin Seeds Dataset",
     "Can we predict the pumpkin seed variety (Çerçevelik vs. Ürgüp Sivrisi) based on morphology?",
     "Logistic Regression works well for binary classification, providing coefficients that indicate feature importance and decision boundaries."),
    
    ("k-Nearest Neighbors (kNN)",
     "Iris Flower Dataset",
     "Which species does a flower belong to given petal and sepal dimensions?",
     "kNN classifies by comparing the sample to its closest neighbors in feature space, making it intuitive for small datasets."),
    
    ("Decision Trees",
     "Titanic Dataset",
     "Which passengers are likely to survive based on demographics and ticket class?",
     "Decision Trees partition data by feature thresholds, creating human-readable rules for classification."),
    
    ("Random Forest",
     "Breast Cancer Dataset (sklearn)",
     "Is a tumor malignant or benign based on medical imaging features?",
     "Random Forest builds multiple decision trees and averages their predictions, improving generalization and reducing overfitting."),
    
    ("Support Vector Machines (SVM)",
     "Handwritten Digits Dataset (sklearn)",
     "What digit (0–9) is represented in a given image?",
     "SVM finds the optimal hyperplane in high-dimensional space, effective for image and text classification tasks.")
]

for alg, data, q, exp in rows:
    row_cells = table.add_row().cells
    row_cells[0].text = alg
    row_cells[1].text = data
    row_cells[2].text = q
    row_cells[3].text = exp

# %%
# Save document
output_path = "classification_scavenger_hunt.docx"
doc.save(output_path)

print(f"Document saved as {output_path}")
```
