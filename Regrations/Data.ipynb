```python
# Exploring Pumpkin Data Visualizations
# Comparing Matplotlib vs Seaborn

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set the style for better looking plots
plt.style.use('default')
sns.set_theme(style="whitegrid")

# Load the pumpkin data
# Note: This assumes we have a CSV file with pumpkin data
# For demonstration, I'll create some sample data
np.random.seed(42)

# Create sample pumpkin data
data = {
    'weight': np.random.normal(8, 2, 200),
    'diameter': np.random.normal(10, 3, 200),
    'color': np.random.choice(['Orange', 'White', 'Green', 'Yellow'], 200, p=[0.6, 0.2, 0.15, 0.05]),
    'variety': np.random.choice(['Howden', 'Cinderella', 'Jarrahdale', 'Baby Boo'], 200),
    'price': np.random.uniform(2, 10, 200)
}

df = pd.DataFrame(data)
df['weight'] = df['weight'].clip(lower=0)  # Ensure no negative weights
df['diameter'] = df['diameter'].clip(lower=0)  # Ensure no negative diameters

print("Pumpkin Dataset Overview:")
print(df.head())
print(f"\nDataset shape: {df.shape}")
```

```python
# Matplotlib implementation
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(df['weight'], bins=15, alpha=0.7, color='orange', edgecolor='black')
plt.title('Matplotlib: Pumpkin Weight Distribution')
plt.xlabel('Weight (lbs)')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Seaborn implementation
plt.subplot(1, 2, 2)
sns.histplot(data=df, x='weight', bins=15, kde=True, color='orange')
plt.title('Seaborn: Pumpkin Weight Distribution')
plt.xlabel('Weight (lbs)')

plt.tight_layout()
plt.show()
```

```python
# Matplotlib implementation
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
colors = {'Orange': 'orange', 'White': 'lightgray', 'Green': 'green', 'Yellow': 'yellow'}
for color in df['color'].unique():
    subset = df[df['color'] == color]
    plt.scatter(subset['diameter'], subset['weight'], 
                alpha=0.7, label=color, c=colors[color])
plt.title('Matplotlib: Weight vs Diameter by Color')
plt.xlabel('Diameter (inches)')
plt.ylabel('Weight (lbs)')
plt.legend()
plt.grid(True, alpha=0.3)

# Seaborn implementation
plt.subplot(1, 2, 2)
sns.scatterplot(data=df, x='diameter', y='weight', hue='color', 
                palette=colors, alpha=0.7)
plt.title('Seaborn: Weight vs Diameter by Color')
plt.xlabel('Diameter (inches)')
plt.ylabel('Weight (lbs)')

plt.tight_layout()
plt.show()
```

```python
# Matplotlib implementation
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
varieties = df['variety'].unique()
box_data = [df[df['variety'] == var]['price'].values for var in varieties]
plt.boxplot(box_data, labels=varieties)
plt.title('Matplotlib: Price Distribution by Variety')
plt.xlabel('Pumpkin Variety')
plt.ylabel('Price ($)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Seaborn implementation
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='variety', y='price')
plt.title('Seaborn: Price Distribution by Variety')
plt.xlabel('Pumpkin Variety')
plt.ylabel('Price ($)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# First, let's create a cross-tabulation
cross_tab = pd.crosstab(df['variety'], df['color'])

# Matplotlib implementation
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
x = np.arange(len(cross_tab.index))
width = 0.2
colors = ['orange', 'lightgray', 'green', 'yellow']

for i, color in enumerate(cross_tab.columns):
    plt.bar(x + i*width, cross_tab[color], width, label=color, color=colors[i])

plt.xlabel('Variety')
plt.ylabel('Count')
plt.title('Matplotlib: Pumpkin Count by Variety and Color')
plt.xticks(x + width*1.5, cross_tab.index)
plt.legend()
plt.grid(True, alpha=0.3)

# Seaborn implementation
plt.subplot(1, 2, 2)
cross_tab_melted = cross_tab.reset_index().melt(id_vars='variety', 
                                                var_name='color', 
                                                value_name='count')
sns.barplot(data=cross_tab_melted, x='variety', y='count', hue='color', 
            palette=colors)
plt.title('Seaborn: Pumpkin Count by Variety and Color')
plt.xlabel('Variety')
plt.ylabel('Count')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
