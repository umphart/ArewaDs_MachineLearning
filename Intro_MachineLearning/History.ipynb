```python
import plotly.express as px
import pandas as pd
```

```python
# %%


# Timeline data
data = [
    {"Year": 1943, "Event": "McCulloch & Pitts Neuron", 
     "Details": "First mathematical model of a neuron.", 
     "Media": "https://upload.wikimedia.org/wikipedia/commons/thumb/e/e3/Neuron3.png/200px-Neuron3.png"},
    
    {"Year": 1958, "Event": "Perceptron (Rosenblatt)", 
     "Details": "Frank Rosenblatt builds the perceptron algorithm.", 
     "Media": "https://upload.wikimedia.org/wikipedia/commons/thumb/5/5c/Frank_Rosenblatt_1960s.jpg/220px-Frank_Rosenblatt_1960s.jpg"},
    
    {"Year": 1986, "Event": "Backpropagation Popularized", 
     "Details": "Rumelhart, Hinton, and Williams bring backpropagation to wide use.", 
     "Media": "https://upload.wikimedia.org/wikipedia/commons/7/7e/Backpropagation_network.png"},
    
    {"Year": 1997, "Event": "Deep Blue Beats Kasparov", 
     "Details": "IBM’s Deep Blue defeats Garry Kasparov in chess.", 
     "Media": "https://upload.wikimedia.org/wikipedia/commons/6/6f/Deep_Blue.jpg"},
    
    {"Year": 2012, "Event": "AlexNet Wins ImageNet", 
     "Details": "Deep CNNs outperform traditional methods, sparking the deep learning boom.", 
     "Media": "https://upload.wikimedia.org/wikipedia/commons/thumb/6/6a/ImageNet_logo.jpeg/200px-ImageNet_logo.jpeg"},
    
    {"Year": 2016, "Event": "AlphaGo Defeats Lee Sedol", 
     "Details": "DeepMind’s AlphaGo beats world champion Go player.", 
     "Media": "https://upload.wikimedia.org/wikipedia/commons/thumb/b/be/AlphaGo_Lee_Sedol.jpg/250px-AlphaGo_Lee_Sedol.jpg"}
]

df = pd.DataFrame(data)

# Create timeline
fig = px.timeline(
    df, 
    x_start="Year", x_end="Year", 
    y="Event", 
    hover_data=["Details", "Media"],
    color="Year"
)

fig.update_yaxes(autorange="reversed")  # Keep chronological order
fig.show()
```
